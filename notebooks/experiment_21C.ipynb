# Experiment 21C

Notebook ini membaca artefak hasil `python -m src.experiment_21C.run_experiment_21C`. Jalankan pipeline Python terlebih dahulu.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
RESULTS = ROOT / 'results' / 'experiment_21C'
required = [
    RESULTS / 'experiment_21C_summary.json',
    RESULTS / 'experiment_21C_table.csv',
    RESULTS / 'side_history_exp21C.json',
    RESULTS / 'fusion_predictions_exp21C.csv',
]
missing = [str(path) for path in required if not path.exists()]
if missing:
    raise FileNotFoundError('Artefak belum lengkap. Jalankan: python -m src.experiment_21C.run_experiment_21C\n' + '\n'.join(missing))

summary = json.loads((RESULTS / 'experiment_21C_summary.json').read_text(encoding='utf-8'))
table = pd.read_csv(RESULTS / 'experiment_21C_table.csv')
side_history = json.loads((RESULTS / 'side_history_exp21C.json').read_text(encoding='utf-8'))
fusion_predictions = pd.read_csv(RESULTS / 'fusion_predictions_exp21C.csv')
table
```

```python
print(summary['description'])
print('\nExp 21B reference:', summary['reference_exp21B'])
print('Final v1 reference:', summary['reference_final_v1'])
```

```python
epochs = range(1, len(side_history['train_loss']) + 1)
fig, ax1 = plt.subplots(figsize=(9, 4))
ax1.plot(epochs, side_history['train_loss'], label='train loss')
ax1.plot(epochs, side_history['val_loss'], label='val loss')
ax1.set_xlabel('epoch')
ax1.set_ylabel('loss')
ax2 = ax1.twinx()
ax2.plot(epochs, side_history['val_macro_f1'], color='tab:green', label='val Macro F1')
ax2.set_ylabel('Macro F1')
lines, labels = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines + lines2, labels + labels2, loc='lower right')
ax1.set_title('Experiment 21C Side')
ax1.grid(alpha=0.25)
plt.show()
```

```python
display_cols = ['method', 'accuracy', 'precision_macro', 'recall_macro', 'f1_macro', 'safe_to_phone', 'phone_to_safe', 'generalization_status']
table[display_cols]
```

```python
ax = table.set_index('method')['f1_macro'].plot(kind='bar', figsize=(8, 4), color=['#3b6ea8', '#879957', '#c05a3b', '#7b5fa8'])
ax.axhline(summary['reference_exp21B']['average_fusion_f1_macro'], color='gray', linestyle=':', linewidth=1, label='Exp 21B fusion F1')
ax.axhline(summary['reference_final_v1']['average_fusion_f1_macro'], color='black', linestyle='--', linewidth=1, label='Final v1 fusion F1')
ax.set_ylabel('Macro F1')
ax.set_title('Experiment 21C Macro F1')
ax.legend()
plt.xticks(rotation=25, ha='right')
plt.tight_layout()
plt.show()
```
